```python
#list creation
numbers = [10, 20, 30, 40, 50]
print(numbers)
```

```python
#accesing elements
print("First element:", numbers[0])
print("Last element:", numbers[-1])
```

```python
#slicing
print(numbers[1:3])
print(numbers[:2])
print(numbers[2:])

```

```python
#modifying
numbers[1] = 25
print(numbers)
```
