```python
import sqlite3
import os
# Import the os module
# Connect to SQLite database
# If example.db exists,
if os.path.exists("example.db"):
    os.remove("example.db")
    print("🗑️ Existing 'example.db' removed to ensure a clean start.")

conn = sqlite3.connect("example.db")
cursor = conn.cursor()
# Create Table
cursor.execute("""
CREATE TABLE IF NOT EXISTS users (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    age INTEGER,
    email TEXT UNIQUE
)
""")
print("✅ Table created successfully.")
# Insert Data (Safe Insert)
users = [
    ("Alice", 25, "alice@example.com"),
    ("Bob", 30, "bob@example.com")
]

for user in users:
    try:
        cursor.execute(
            "INSERT INTO users (name, age, email) VALUES (?, ?, ?)",
            user
        )
    except sqlite3.IntegrityError:
        print(f"⚠️ User with email {user[2]} already exists.")

conn.commit()
print("✅ Data inserted successfully.")

# -------------------------------
# Read Data (SELECT)
# -------------------------------
print("\n📄 Users in database:")
cursor.execute("SELECT * FROM users")
rows = cursor.fetchall()

for row in rows:
    print(f"ID:{row[0]} | Name:{row[1]} | Age:{row[2]} | Email:{row[3]}")
# Update Data
cursor.execute(
    "UPDATE users SET age = ? WHERE name = ?",
    (28, "Alice")
)
conn.commit()
print("\n✅ Record updated successfully.")
# Delete Data
cursor.execute(
    "DELETE FROM users WHERE name = ?",
    ("Bob",)
)
conn.commit()
print("✅ Record deleted successfully.")
# Display Final Data
print("\n📄 Final data:")
cursor.execute("SELECT * FROM users")
rows = cursor.fetchall()
for row in rows:
    print(f"ID:{row[0]} | Name:{row[1]} | Age:{row[2]} | Email:{row[3]}")
# Close Connection
conn.close()
print("\n✅ Database connection closed.")
```
