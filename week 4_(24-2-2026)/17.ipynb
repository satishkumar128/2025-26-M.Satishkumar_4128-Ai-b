```python
import pandas as pd
import numpy as np
# Create sample data
df = pd.DataFrame({
    'A': [10, 20, np.nan, 40],
    'B': [5, np.nan, 15, 20]
})
# a) Check missing values
print("Missing values:\n")
print(df.isna())
# b) Sum of missing values
print("\nTotal missing values in each column:")
print(df.isna().sum())
# c) Replace missing values with column mean
df = df.fillna(df.mean())
print("\nData after replacing missing values with mean:")
print(df)
```
