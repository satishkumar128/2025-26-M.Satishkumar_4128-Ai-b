```python
# filtering data
import pandas as pd

data = {
    "ID": [101, 102, 103, 104, 105],
    "Name": ["Anu", "Bobu", "Charlie", "Dravid", "Iva"],
    "Age": [21, 25, 23, 24, 28],
    "Marks": [86, 78, 92, 89, 93]
}
df = pd.DataFrame(data)
print("Og Dataframe:")
print(df)
print("\n Students with marks >85:")
print(df[df["Marks"] > 85])
print("\n students with age <25:")
print(df[df["Age"] < 25])
print(df[(df["Age"] > 23) & (df["Marks"] > 88)])
print(df[df["Marks"] > 90], df["Age"] > 23)
```

```python
# Data Manipulation
import pandas as pd
data = {
    "Name": ["A", "B", "C"],
    "Age": [20, 21, 22],
    "Marks": [70, 80, 90]
}
df = pd.DataFrame(data)
print("Orig data:")
print(df)
df["Bonus"] = 5
print("\n after add Bonus:")
print(df)
df["Marks"] = df["Marks"] + 2
print("\n after increas marks")
print(df)
df.loc[0, "Age"] = 21
print("\n after changing age of first stud:")
print(df)
df.rename(columns={"total": "final"}, inplace=True)
print("\n After renaming final:")
print(df)
df.drop("Bonus", axis=1, inplace=True)
print("\n After removing Bonus:")
print(df)
```

```python
# Ranking

import pandas as pd

data = {
    "Name": ["A", "B", "C", "D"],
    "Marks": [85, 92, 74, 82]
}
df = pd.DataFrame(data)
print("Original Data:")
print(df)
df["Rank"] = df["Marks"].rank(ascending=False)
print("\n after ranking:")
print(df)

```

```python
#mean,meadiam,mode,standard variation,variance,kurtosis,skewness
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
data = [10, 12, 15, 18, 20, 22, 25, 30, 35, 40]
df = pd.DataFrame(data, columns=["values"])
print("Dataset:")
print(df)
print("\nSummary statistics:")
print(df.describe())
print("\nvariance:", np.var(data))
print("Standard deviation:", np.std(data))
plt.hist(data)
plt.xlabel("values")
plt.ylabel("frequency")
plt.title("data Distribution")
plt.show()
mean_value = np.mean(data)
median_value = np.median(data)
print("\ninference")
print("mean:", mean_value)
print("median:", median_value)
if mean_value > median_value:
    print("Data is right-skewed")
elif mean_value < median_value:
    print("Data is left-skewed")
else:
    print("Data is symmetric")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
url = "https://raw.githubusercontent.com/dataprofessor/data/master/BostonHousing.csv"
df = pd.read_csv(url)
print("Dataset (first 10 rows):")
print(df.head(10))
data = df["medv"]
print("In summary statistics:")
print(data.describe())

print("In Variance:", np.var(data))
print("Standard deviation:", np.std(data))
print("Kurtosis (Pandas):", data.kurt())
print("Kurtosis (Scipy):", kurtosis(data))

plt.hist(data, bins=10, color='skyblue', edgecolor='black')
plt.axvline(np.mean(data), color='red',
            linestyle='dashed',
            linewidth=2,
            label=f"Mean: {np.mean(data):.2f}")
plt.axvline(np.median(data), color='green',
            linestyle='dashed',
            linewidth=2,
            label=f"Median: {np.median(data):.2f}")
plt.xlabel("Median Home Value (medv)")
plt.ylabel("frequency")
plt.title("Distribution of Median Home Value")
plt.legend()
plt.show()
mean_value = np.mean(data)
median_value = np.median(data)
print("In Inter_Inference:")
print("Mean value:", mean_value)
print("Median value:", median_value)
if mean_value > median_value:
    print("Data is slightly right_skewed")
elif mean_value < median_value:
    print("Data is slightly left-skewed")
else:
    print("Data is symmetric")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import skew, kurtosis
url = "https://raw.githubusercontent.com/dataprofessor/data/master/BostonHousing.csv"
df=pd.read_csv(url)
print("dataset of (first 10 rows):")
print(df.head(10))
print("\n summary statistics of :")
print(df.describe())
print("\nVariance for each col")
print(df.var())
print("skewness")
print(df.apply(lambda x: skew(x.dropna())))
print("\nkurtosis")
print(df.apply(lambda x: kurtosis(x.dropna())))
stats_df=pd.DataFrame({
    "count": df.count(),
    "mean": df.mean(),
    "std": df.std(),
    "variance": df.var(),
    "min": df.min(),
    "median": df.median(),
    "max": df.max(),
    "skewness": df.apply(lambda x: skew(x.dropna())),
    "kurtosis": df.apply(lambda x: kurtosis(x.dropna()))
})
print("\n combined descriptive statistics table:")
print(stats_df)

```

```python
# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"
df = pd.read_csv(url)
print("First 5 rows of dataset:")
print(df.head())
print("\nDataset Info:")
print(df.info())
plt.figure()
plt.plot(df['sepal_length'])
plt.title("sepal length line Plot")
plt.xlabel("index")
plt.ylabel("sepal length")
plt.show()
plt.figure()
plt.hist(df['petal_length'], bins=20)
plt.title("petal length Distribution")
plt.xlabel("petal length")
plt.ylabel("frequency")
plt.show()
plt.figure()
plt.scatter(df['sepal_length'], df['petal_length'])
plt.title("sepal Length vs Petal Length")
plt.xlabel("sepal Length")
plt.ylabel("petal Length")
plt.show()
plt.figure()
df['species'].value_counts().plot(kind='bar')
plt.title("species Count")
plt.xlabel("Species")
plt.ylabel("Count")
plt.show()
plt.figure()
species_count = df['species'].value_counts()

plt.pie(species_count,
        labels=species_count.index,
        autopct='%1.1f%%',
        startangle=90)
plt.title("Species distribution pie Chart")
plt.axis('equal')
plt.show()
```
