25). Write a code to implement logistic regression to find the following

A) spam or not

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.linear_model import LinearRegression, LogisticRegression
data={
    "message":[
        "Win money now","Hello friend","Free prize claim", "How are you","Congratulations you won"],
    "label":[1,0,1,0,1]
}
df_spam = pd.DataFrame(data)
vectorizer=CountVectorizer()
X=vectorizer.fit_transform(df_spam["message"])
Y=df_spam["label"]
model=LogisticRegression()
model.fit(X,Y)
test=vectorizer.transform(["free money offer"])
print("Spam prediction:",model.predict(test))
```

B) disease or not disease

```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
data={
    "age":[25,45,356,50,23],
    "bp":[120,140,130,150,110],
    "disease":[0,1,1,1,0]
}
df=pd.DataFrame(data)
X=df[["age","bp"]]
Y=df["disease"]
model=LogisticRegression()
model.fit(X,Y)
print("Disease prediction:", model.predict([[40,135]]))
```

C) sentiment analysis

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.linear_model import LogisticRegression
data={
    "Review":[
        "I love this product",
        "This is bad",
        "Amazing experience",
        "Worst ever",
        "Very good quality"
    ],
    "sentinment":[1,0,1,0,1]
}
df=pd.DataFrame(data)
vectorizer=CountVectorizer()
X=vectorizer.fit_transform(df["Review"])
Y=df["sentinment"]
model=LogisticRegression()
model.fit(X,Y)
test=vectorizer.transform(["this product is good"])
print("Sentiment prediction:", model.predict(test))
```
